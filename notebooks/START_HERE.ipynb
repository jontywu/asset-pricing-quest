## 这门课要回答的问题
为什么有的股票长期平均收益高、有的低？资产定价的基本回答是：**平均收益是对承担风险的补偿**。一项资产如果恰好在经济不好、大家都缺钱的时候亏损，投资者就要求它平时提供更高的回报，否则不愿意持有。难点在于：到底是哪些风险获得补偿，怎样用数据检验？

这套课程沿着学术史，把这个回答从提出、检验到修正的过程亲手做一遍：
1. **CAPM（资本资产定价模型）** 认为只有一种风险获得补偿：资产随整个股市一起涨跌的程度，用**市场 β** 衡量。
2. 数据显示，小公司股票和"便宜"股票（账面价值相对市值高的**价值股**）的平均收益高于 CAPM 的预测。这类现有模型解释不了的收益规律称为**异象**。
3. Fama 和 French（1993）把"规模"和"价值"做成两个**因子**（做多一类股票、做空另一类股票的组合收益），与市场一起组成**三因子模型**；2015 年再加入盈利和投资，成为**五因子模型**。
4. 最后检验这些模型换一个时期、换一批资产是否依然有效。

每关都按同一结构展开：**经济问题 → 需要的概念 → 数据与代码 → 读结果 → 结论能推到哪里 → 自己作答**。

## 数据从哪里来
全部股票数据来自 Kenneth French 教授的公开数据库 [Data Library](https://mba.tuck.dartmouth.edu/pages/faculty/ken.french/data_library.html)。French 是 Fama–French 模型的作者之一，他按论文规则把美国股票整理成组合，按月公布这些组合和因子的收益。我们直接使用这些整理好的月度收益，所以不需要个股数据库（如 WRDS 上的 CRSP、Compustat）也能完成全部课程。第 04 关会用一组虚构的小例子，演示个股怎样被分组、加权成组合。
